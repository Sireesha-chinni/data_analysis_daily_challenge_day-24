# Coding Challenge: Ride Booking Analysis


You are working as a data analyst for a ride-booking company like Ola/Uber.



Each ride record is stored as a tuple:


(ride_id, driver_name, city, fare, rating)

```python
rides = (
(101, "Ravi", "Bangalore", 250, 4.5),
(102, "Anjali", "Hyderabad", 180, 4.2),
(103, "Kiran", "Bangalore", 300, 4.8),
(104, "Meena", "Chennai", 220, 4.0),
(105, "Arjun", "Bangalore", 150, 3.9)
)
```

```python
rides
```

Tasks

Count total rides


Find how many ride records are present.

```python
print("Total ride records:", len(rides))
```

Extract all rides from Bangalore



Print only tuples where city = "Bangalore"

```python
city="Bangalore"

for i in range(len(rides)):
    if city in rides[i]:
        print(rides[i])
```

Find total revenue from Bangalore


Sum of fares where city is Bangalore

```python
city="Bangalore"
tot_rev = 0

for i in range(len(rides)):
    if city in rides[i]:
        tot_rev = tot_rev + rides[i][3]

print("Sum of fares where city is Bangalore =", tot_rev)
```

Find highest rated driver



Print the tuple with maximum rating

```python
t=[]

for i in range(len(rides)):
    t.append(rides[i][4])
    
print(tuple(t))
print("Highest rated driver:",rides[t.index(max(t))])
```

Average rating per city



Output should look like:
{
"Bangalore": 4.4,
"Hyderabad": 4.2,
"Chennai": 4.0
}

```python
rides = (
    (101, "Ravi", "Bangalore", 250, 4.5),
    (102, "Anjali", "Hyderabad", 180, 4.2),
    (103, "Kiran", "Bangalore", 300, 4.8),
    (104, "Meena", "Chennai", 220, 4.0),
    (105, "Arjun", "Bangalore", 150, 3.9)
)

totals = {}   # city -> sum of ratings
counts = {}   # city -> number of rides

for ride in rides:
    city = ride[2]
    rating = ride[4]
    totals[city] = totals.get(city, 0) + rating
    counts[city] = counts.get(city, 0) + 1

avg_rating = {city: round(totals[city] / counts[city], 1) for city in totals}

print(avg_rating)
```
